### Q1: NumPy Array Broadcasting


```python
import numpy as np

a = np.array([[1], [2], [3]])
b = np.array([[10, 20, 30]])

result = a + b

print("Result:")
print(result)
print("Shape:", result.shape)

b = np.array([1, 2])

try:
    print(a + b)
except ValueError as e:
    print("Error:", e)
```

**Explanation:** NumPy broadcasting allows arrays with compatible shapes to perform element-wise addition. Arrays with shapes (3,1) and (1,3) produce a (3,3) result. Replacing `b` with an array of shape (2,) causes a `ValueError` because the shapes are incompatible.


### Q2: Handling CSV Data Issues


```python
import pandas as pd
from io import StringIO

data = StringIO("""ID,Name,Date,Marks
101,"Asha, R",2026-09-01,80
102,Ravi,01/09/2026,
103,Neha,2026-09-03,NA
""")

df = pd.read_csv(data, na_values=["NA"])

df["Date"] = pd.to_datetime(
    df["Date"], dayfirst=True, errors="coerce"
)
df["Marks"] = pd.to_numeric(
    df["Marks"], errors="coerce"
)

print(df)
print(df.dtypes)
```

**Explanation:** Quotation marks allow commas inside names to be read correctly. `to_datetime()` converts the Date column into datetime format, and `to_numeric()` converts Marks into numeric values. Missing values are represented as NaN. The meaning of `NA` should be checked before treating it as missing data.


### Q3: Mean, Median, and Outliers


```python
import pandas as pd

df = pd.DataFrame({
    'Salary': [25000, 30000, 35000, 10000000]
})

print("Mean:", df['Salary'].mean())
print("Median:", df['Salary'].median())

print("Highest salary record:")
print(df.loc[df['Salary'].idxmax()])
```

**Explanation:** The unusually high salary increases the mean significantly, while the median is less affected by extreme values. Therefore, the median better represents a typical salary in this dataset. The extreme value should be investigated before removing it.


# Section D: Scenario-Based Questions

### Q1: Data Cleaning


```python
import pandas as pd

df = pd.DataFrame({
    'Student': ['Asha', 'Ravi', 'Asha', 'Neha'],
    'Department': ['IT', 'it', 'IT', 'I.T.'],
    'Marks': ['80', None, '80', '90']
})

df['Department'] = (
    df['Department'].str.strip().str.upper()
    .replace({'I.T.': 'IT'})
)

df['Marks'] = pd.to_numeric(df['Marks'], errors='coerce')

df = df.drop_duplicates()

df['Marks'] = df['Marks'].fillna(df['Marks'].median())

print(df)
```

**Explanation:** This code standardizes department names, converts Marks into numeric values, removes duplicate records, and fills missing marks using the median. Missing values should be investigated before deleting records because they may have important reasons.


### Step 2: Identify Invalid Marks

```python
invalid_marks = df[(df["Marks"] < 0) | (df["Marks"] > 100)]

print("Invalid Marks:")
print(invalid_marks)
```

### Step 3: Check Missing Values

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

### Step 4: Display the Final DataFrame

```python
print("Final Cleaned DataFrame:")
print(df)
```

### Step 5: Summary of Data Cleaning

```python
print("Total rows:", len(df))
print("Duplicate rows remaining:", df.duplicated().sum())
print("Missing values remaining:", df.isnull().sum().sum())
print("Invalid marks remaining:", ((df["Marks"] < 0) | (df["Marks"] > 100)).sum())
```

### Step 7: Verify the Final Output

```python
print(df)
print("\nData types:")
print(df.dtypes)

print("\nMarks summary:")
print(df["Marks"].describe())
```

### Step 8: Conclusion

Data cleaning improves data quality by standardizing department names, converting marks into numeric values, identifying duplicate records, and checking missing and invalid values. Missing marks should be investigated before filling or removing them to avoid losing important information.
